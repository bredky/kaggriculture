```python
from kaggle_environments import make
from agent_melon import melon_maxxer
from agent_melon_test import test_melon_maxxer

env = make("kaggriculture", configuration={"episodeSteps": 720}, debug=True)
env.run([test_melon_maxxer, test_melon_maxxer])  # or env.run(["main.py", "random"]) to load from a file

# View result
final = env.steps[-1]
for i, s in enumerate(final):
    print(f"Player {i}: reward={s.reward}, status={s.status}")

#render
with open("replay.html", "w") as f:                                                                                              
    f.write(env.render(mode="html", width=1200, height=800))   

# Or dump a replay JSON for the visualizer / offline analysis
import json
with open("replay.json", "w") as f:
    json.dump(env.toJSON(), f)
```
